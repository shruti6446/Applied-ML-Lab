```python
import numpy as np
```

```python
x=np.array([1,2,3,4,5])
y=np.array([35,40,50,55,65])
```

```python
m,c =np.polyfit(x,y,1)
print('Slope:', m)
print('Intercept:', c)
```

```python
y_pred= m * x + c
errors= y-y_pred
mae= np.mean(np.abs(errors))
mse= np.mean(errors**2)
rmse= np.sqrt(mse)

print('Slope:', m)
print('Intercept:', c)
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
```
